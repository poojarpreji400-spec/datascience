```python
#Calculate electricity bill with different slabs 
units = 300  #First 100 units → ₹5 per unit, Next 100 units → ₹7 per unit, Above 200 units → ₹10 per unit
if units<= 100:
    bill= units*5
elif units<=200:
    bill = (100*5) + ((units-100)*7)
else:
    bill = (100*5) + (100*7) + ((units-200)*10)
print('electricity bill:', bill)
```

```python
#Check if a number is palindrome using while loop
num = 121 # Original number and reversed number are equal
temp = num 
rev = 0
while num>0:
    digit = num%10
    rev = rev*10 + digit
    num = num//10
if temp == rev:
    print('palindrome')
else:
    print('not palindrome')
```

```python
#Find all prime factors of a number
num = 20
for i in range(2, num+1):
    if num%i ==0: # only factors goes to the next step Eg:2,4,5,...
        prime = True #Assume factor is prime, then the below loop proves it wrong or right
        for j in range(2,i): # we cannot prove prove 2 is not prime and we alreday said that prime = true, so 2 is prime.
            if i %j == 0:
                prime = False # is it becomes =0, then it is not prime(prime = false) for, i = 4, j = 2,3, since 4%2==0 j loop breaks
                break
        if prime:
            print(i)
```

```python
#Password validation with multiple conditions
special_chars = '#@$%&*'
while True:
    password = input('enter the password:')
    if len(password)<12:
        print('invalid, the password must contain atleat 12 chars')
    if not any(char.isdigit() for char in password):
        print('invalid, the password muts contain atleast one digit')
    if not any(char.isupper() for char in password):
        print('invalid, the password must contain alteast one capital')
    if not any(char in special_chars for char in password):
        print('invalid, the password must contain alteast one special character')
    if (len(password)>=12 and any(char.isdigit() for char in password) and any(char.isupper() for char in password) and any(char in special_chars for char in password)):
        print('valid password')
        break
```
