```python
#5. Write a function to compute factorial of a number.
def factorial(n):
    if n==0:
        return 1
    else:
        return n*factorial(n-1)
factorial(4)       
```

```python
#7. Write a function to check if a string is a palindrome.
def palindrome(n):
    if n == n[::-1]:
        return 'palindrome'
    else:
        return 'not palindrone'
print(palindrome('madam'))
```

```python
#9. Write a function to count vowels in a string.
def count_vowels(str):
    count=0
    vowels = 'aeiouAEIOU'
    for char in str:
        if char in vowels:
            count += 1
    return count
print(count_vowels('ropeis'))
```

```python
#10. Write a function that checks if a number is prime.
def check_prime(n):
     if n<=1:
         return 'not prime'
     for i in range(2,n):
         if n % i ==0:
             return 'not prime'
     return 'prime'
check_prime(4)
```

```python
#11. Write a function to generate Fibonacci series up to n terms.
def fibonacci(n):
    if n == 0:
        return 0
    elif n ==1:
        return 1
    else:
        return fibonacci(n-1) + fibonacci (n-2)
print(fibonacci(7))
```

```python
#19. Write a function that accepts a number and returns the sum of its digits.
def sum_digits(n):
    return sum(int(digit) for digit in str(n)) # the code run backwards, so first convert num to string using str(n) then each digit in string is converted to integer then there sum is taken
print(sum_digits(1234))
```

```python
#20. Write a function to check if a given year is a leap year.
def check_leap_year(n):
    if (n%4 ==0 and n%100 !=0) or (n%400==0):
        return 'leap year' #A year is a leap year if it is divisible by 4, except for end-of-century years (ending in 00), which must also be divisible by 400
    else:
        return 'not leapyear'
print(check_leap_year(2023))
```
