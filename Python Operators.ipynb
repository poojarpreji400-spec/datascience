```python
rows = 5
for i in range(1, rows+1):
    for j in range(i):
        print('*',end =" ")
    print() #Outer loop = number of rows, Inner loop = number of stars in each row
```

```python
rows = 6
for i in range(rows,0,-1):
    for j in range(i):
        print('*', end=' ')
    print()
```

```python
# ATM withdrawal program
balance = 6000
atm_pin = 9867
while True:
    entered_pin = int(input('enter your pin:'))
    if entered_pin == atm_pin:
        print('Pin accepted')
        withdraw = int(input('enter amt:'))
        if withdraw <= balance:
            balance -= withdraw
            print(f'withdrawal successful:,remaing balance:{balance}')
        else:
             print('insuffient balance')
        break
    else:
         print('wrong pin') 
```
