```python
#2. Handle the error when a user enters a non-integer input for age.
try:
    age = int(input('enter a num:'))
    print(f'age:{age}')
except ValueError:
    print('age cannot be non-integer')
```

```python
#3. Write a program that uses try, except, else to divide two numbers.
try:
    a=int(input('enter a num:'))
    b=int(input('enter a num:'))
    result = print(a/b)
except ZeroDivisionError:
    print('a num cannot be divided by 0')
else:
    print(result)
```

```python
#4. Handle FileNotFoundError while reading a non-existing file.
try:
    f= open('happs.txt' ,'r')
    content=f.read()
    print(content)
except FileNotFoundError:
    print('this file does not exit')
```

```python
#5. Create a custom exception called NegativeNumberError and raise it if a negative number is entered.
class NegativeNumberError(Exception):
    pass
try:
    n = int(input('enter a num:'))
    if n <0:
        raise NegativeNumberError('Negative num entered')
except NegativeNumberError as e:
    print(e)
```
