```python
sns.countplot(x = 'feels_burned_out', data = data, color = 'pink')
plt.title('Number of students feeling burned out')
plt.xlabel('feels_burned_out')
plt.ylabel('students_count')
```

1) Out of 3000 students in the datset, 2039 students feeling burned out and 961 are not burned out.
This expalins that majority of the students are feeling burned out.

```python
data['Screen_Time_Group'] = pd.cut(data['screen_time_hours'], bins = [0,2,4,6,8,10], labels = ['0-2', '2-4','4-6', '6-8', '8-10'])
sns.countplot(x='Screen_Time_Group', hue='feels_burned_out', data=data)
plt.title('Time spent on screen vs Burn out')
plt.xlabel('Time spent on screen')
plt.ylabel('Number of students')
plt.show()
```

```python
data['Screen_Time_Group'].value_counts()
```

```python
screen_burnout_percentage = data.groupby('Screen_Time_Group')['feels_burned_out'].mean()*100
screen_burnout_percentage
```

1) As the screentime increases, the burnout also increases.
2) The Barplot shows that large number of students spend for almost 6 - 8 hours on screen time followed by 8 - 10 hours.
3) Since the number of childern in each study group is different(Eg: there are 1045 students who use their screens for 6-8 hours and 558 students who use their screens for for 8-10 hours), the barplot alone cannot be used to determine the burnout.
4) Therefore, the burnout percentage gives a more clear representation. It shows that about 88.53% of students in 8 - 10 hours screen time reported burnout, which is the highest. 

```python
education_burnout = data.groupby('education_level')['feels_burned_out'].value_counts()
education_burnout
```

```python
plt.figure(figsize =(10,6))
education_burnout.plot(kind='bar')
plt.title('Education level VS Burnout')
plt.xlabel('Education level')
plt.ylabel('Number of students') 
plt.xticks(rotation = 45)
plt.show()
```

1) Among 3000 students, most of them are undergraduate students.
2) Undergraduate students are feeling highest burnout compared to Graduate and High School students.
3) High School students have less burnout comapred to Undergraduate.
4) Graduate students have the least burnout, but their burnout out cases are still more than non burn out cases.

```python
data.groupby('exercise_hours_per_week')['feels_burned_out'].value_counts()
```

```python
data['excercise_grouped'] = pd.cut(data['exercise_hours_per_week'], bins =[0,3,6,9,12], labels= ['0-3','3-6','6-9','9-12'])
```

```python
plt.figure(figsize = (8,5))
sns.countplot(x = 'excercise_grouped', hue = 'feels_burned_out', data= data)
plt.title('Exercise VS Burnout')
plt.xlabel('Excercise')
plt.ylabel('number of students')
plt.legend(title = 'Burn Out')
plt.show()
```

```python
avg_gpa = data.groupby('stress_level')['gpa'].mean()
avg_gpa
```

```python
plt.figure(figsize = (8,5))
sns.barplot(x = avg_gpa.index, y = avg_gpa.values)
plt.xlabel('Stress level')
plt.ylabel('avg gpa')
plt.title('Avg Stress VS gpa')
plt.show()
```

1) Students with low stress level have high Gpa.
2) The gpa decreses with increase in stress.
3) Students who have high stress has low gpa.
4) the stress and gpa have negative relationship.
