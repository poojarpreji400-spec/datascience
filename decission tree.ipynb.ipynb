```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
from sklearn.tree import DecisionTreeClassifier
```

```python
data = pd.read_csv('titanic.csv')
data
```

```python
print(data.head())
```

```python
print(data.info())
```

```python
print(data.isnull().sum())
```

```python
data['Age'] = data['Age'].fillna(data['Age'].median())
```

```python
data['Embarked'] = data['Embarked'].fillna(data['Embarked'].mode()[0]) # 0 is uded to get the first value from the mode result
```

```python
data.drop(['PassengerId', 'Name', 'Ticket', 'Cabin'], axis = 1, inplace = True)
```

```python
le_sex = LabelEncoder()
data['Sex'] = le_sex.fit_transform(data['Sex'])
```

```python
le_em = LabelEncoder()
data['Embarked'] = le_em.fit_transform(data['Embarked'])
```

```python
#creating a family size feature
data['Family_size'] = data['SibSp'] +data['Parch'] +1
#the +1 is there because if the sibling count = 3, and parents count = 2,then we have to add the passenger(the passenger is not in SibSp and Parch) also so for that we have+1
```

```python
x = data.drop('Survived', axis = 1)
y = data['Survived']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.2, random_state = 42)
```

```python
model = DecisionTreeClassifier(random_state = 42, max_depth = 5) # here we are giving depth = 5, so that the tree will not grow very complicated, if the tree grows too much it can llearn the datset instead of studying the pattern and cause underfitting
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
print(accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```
