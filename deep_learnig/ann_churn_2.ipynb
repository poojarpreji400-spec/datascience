```python
import pandas as pd 
import numpy as np 
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from sklearn.metrics import classification_report, confusion_matrix
```

```python
data = pd.read_csv('https://raw.githubusercontent.com/sharmaroshan/Churn-Modelling-Dataset/master/Churn_Modelling.csv')
data
```

```python
# Exited is the target column
# 0 --> did not leave, 1--> customer left
data['Exited'].value_counts()
```

```python
# preprocessing
data.drop(['RowNumber', 'CustomerId','Surname'], axis = 1, inplace = True)
```

```python
data.dtypes
```

```python
for column in data.columns:
    if data[column].dtype == object:
        le = LabelEncoder()
        data[column] = le.fit_transform(data[column])
```

* if we want we can one hot (get_dummies) to make many columns, so ann can learn form multiple columns

```python
x = data.drop('Exited', axis = 1)
y = data['Exited']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.2, random_state = 42)
```

```python
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
# Build the ann model
model = Sequential()
model.add(Dense(16, activation = 'relu', input_dim = x_train_scaled.shape[1],))
model.add(Dropout(0.3))
model.add(Dense(8, activation = 'relu'))
model.add(Dropout(0.3))
model.add(Dense(1, activation = 'sigmoid'))
model.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
```

```python
history = model.fit(x_train_scaled, y_train,epochs = 50, validation_split=0.2, verbose = 0)
```

```python
loss, accuracy = model.evaluate(x_test_scaled, y_test)
print('Accuracy is:', accuracy)
```

```python
y_pred = (model.predict(x_test_scaled) >0.5).astype('int32')
print('classification report', classification_report(y_test, y_pred))
print('confusion matrix', confusion_matrix(y_test, y_pred))
```
