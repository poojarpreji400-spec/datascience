```python
from ultralytics import YOLO # You Only Look Once, It is used to detect phone, person, cat, dog, bicycle, etc.
import cv2 # OpenCV is used to acess the webcam, read video frames, display the video, etc.
```

```python
model = YOLO('yolov8n.pt') # bringing the pretrained model object detection brain into my Python program.
```

```python
cap = cv2.VideoCapture(0) # open the camera 0 - use the deafult/ built in webcam.as_integer_ratio# 1 --> if second camera is available
```

```python
while True:   # keep doing the steps again and again
    ret, frame = cap.read() # captures one frame from the output, 
    # there are 2 output 1) ret - tells us whether the frame was successfully captured. (ret = True--> Image Captured successfully, ret= False --> Something went wrong)
    # frame - actual image from the webcame
    
    if not ret:
        break # means if we get ret = False(like i could not get a webcame frame), then stop the loop.    
    results = model(frame) # the result contains the percentage of confident the model has on predicting the object in the frame
    
    annoted_frame = results[0].plot() # This takes the YOLO detection results and draws them on the image.
    # before annotation
    #        👩       🐶
    # After annotation  
    # ┌─────────────┐           ┌─────────────┐   
    # │ person 95%  │           │ dog 91%     │
    # │      👩    │           │      🐶     │
    # └─────────────┘           └─────────────┘
    
    
    cv2.imshow('YOLOv8 Detection', annoted_frame) # this creates a window showing the processed webcam video.
    # the window title is YOLOv8 Detection
    # the image displayed is annoted_frame.
    
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break # If the user presses q, stop the webcam loop.
    
cap.release() # After you exit the loop, this disconnects/releases the webcam., we don't want the camers to be running.
cv2.destroyAllWindows() # This closes the window created by:cv2.imshow()
```
