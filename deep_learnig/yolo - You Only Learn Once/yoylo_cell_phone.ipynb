```python
from ultralytics import YOLO 
import cv2 # webcam
```

```python
model = YOLO('yolov8n.pt')
```

```python
# open the webcam
cap = cv2.VideoCapture(0)
```

```python
while True:
    ret, frame = cap.read()
    # frame - the image, ret - did we successfully got the image or not?
    
    if not ret:
        break
    
    results = model(frame)
    
    for result in results:
        boxes = result.boxes
        # The box tells us where the object is in the image.
        # ┌──────────────┐
        # │    📱        │
        # │              │
        # └──────────────┘
        
        for box in boxes: # the loop process the things detected like person, dog, laptop as box 1---> person, box2---> dog, etc.
            cls_id = int(box.cls[0]) # Get the numerical ID of the detected object's class.
            # Yolo gives each object a class number, like 0--> person, 1--> dog, etc.
            class_name = model.names[cls_id]
            # Eg: cls_id = 67
            #       ↓
            # model.names[67]
            #       ↓
            # "cell phone"
            
            if class_name == 'cell phone': # checking whether it is cell phone
                x1, y1, x2, y2 = map(int, box.xyxy[0]) # gives coordinates of the box containing the cell phone.
                # cordinates may comes as floating point, so int is used to convert it to integer.
                
                confidence = float(box.conf[0]) # tells us how confident the detcted item is in float value.
                
                cv2.rectangle(frame, (x1, y1),(x2, y2), 
                              (0, 255, 0), 2) # drwas a reactangle over the phone.
                # cv2.rectangle(image,top-left,bottom-right,color,thickness)
                
                label = f'Cell Phone {confidence:.2f}'
                
                # display the label
                cv2.putText(frame, label, (x1, y1 -10), # (x1, y1-10) --> position of the  bounding box, it is slightly above the position of the text
                            cv2.FONT_HERSHEY_COMPLEX, # font style
                            0.7, # font size
                            (0,255,0), # color
                            2) # text thickness
    cv2.imshow('cell Phone Detection', frame)
    
    if cv2.waitKey(1) & 0xFF ==ord('q'):
        break
    
cap.release()
cv2.destroyAllWindows()
            
```
