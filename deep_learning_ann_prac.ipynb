```python
import numpy as np 
import pandas as pd 
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
```

**numpy and pandas**: For data manipulation.

**fetch_california_housing**: Loads the dataset.

**train_test_split**: Splits data into training and testing sets.

**StandardScaler**: Normalizes features to improve ANN performance.

**Sequential, Dense**: To build the ANN layers.

**Adam**: Optimizer to update weights during training.

```python
housing = fetch_california_housing()
x = pd.DataFrame(housing.data, columns = housing.feature_names)
y = housing.target # here x is converted to dataframe because it has many columns, but it is no necessary fro y
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

```python
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.fit_transform(x_test)
```

```python
model = Sequential([Input(shape = (x_train_scaled.shape[1],)), # input layer
                    Dense(64, activation = 'relu'), Dense(32, activation = 'relu'), Dense(16, activation = 'relu'), # hidden layers
                    Dense(1)]) # output layer
```

Sequential model: Layers are stacked one after another.

Dense layers: Fully connected layers.

64 → 32 → 16 neurons in hidden layers.

relu activation introduces non-linearity.

Output layer: Single neuron for predicting a continuous value.

```python
# compile
from sklearn.metrics import mean_absolute_error, mean_squared_error
model.compile(optimizer = Adam(learning_rate = 0.001), loss = 'mse', metrics = ['mae'])
```

```python
# train the model
history = model.fit(x_train_scaled, y_train, validation_split=0.2, epochs = 50, batch_size = 32)
```

validation_split=0.2: 20% of training data used to validate model during training.

epochs=50: Model sees the entire training set 50 times.

batch_size=32: Number of samples per weight update.

```python
# evaluate the model
loss, mae = model.evaluate(x_test_scaled, y_test)
print(f'Test MAE:{mae}')
```

Test MAE: 1.26 → On average, the predicted house prices differ from the actual prices by 1.262 units (in the dataset’s scale).

Model performance: The ANN predicts house prices reasonably well, capturing the relationship between features and prices.

```python
# make predictions
predictions = model.predict(x_test_scaled)
print('first 2 predictions:', predictions[:5])
```
