```python
import pandas as  pd 
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split, GridSearchCV
#from sklearn.preprocessing import StandardScaler # scaling - is not needed in tree based models
from sklearn.ensemble import GradientBoostingRegressor
from sklearn. metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import RandomizedSearchCV
```

```python
df = fetch_california_housing()
data = pd.DataFrame(df.data, columns = df.feature_names)
```

```python
data['target'] = df.target
data
```

```python
print(data.info())
```

```python
print(data.isnull().sum())
```

```python
x = data.drop('target', axis = 1)
y =  data['target']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.2, random_state = 42)
```

scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.fit_transform(x_test)

```python
gb = GradientBoostingRegressor(random_state = 42)
gb.fit(x_train, y_train)
```

```python
y_pred = gb.predict(x_test)
```

```python
print('mean_absolute_error', mean_absolute_error(y_test,y_pred))
print('mean squared error', mean_squared_error(y_test, y_pred))
print('r2 score', r2_score(y_test, y_pred))
```


The Baseline Model achieved an R² Score of 0.775, meaning it can explain about 77.6% of the variance in house prices. While this is a decent starting point, it also means that 22.4% of the variance remains unexplained. Since the model's accuracy (R2) is not close to the ideal value of 1.0, there is significant room for improvement. Therefore, we proceed with hyperparameter tuning to find a better configuration that can maximize the predictive performance.

Hyper Parameter Tuning

```python
param_grid = {'n_estimators' : [100,200], 'learning_rate' : [0.1], 'max_depth' : [3,4]}
# n_estimators = determines the number of trees(boosting stages) to use in the ensemble. Here the model will be tested with both 100 and 200 treess.
# learning_rate = controsl how much each tree can contribute for the correction.
#Tree 1 → prediction
#          ↓
#Tree 2 → corrects mistakes
#         ↓
#Tree 3 → corrects remaining mistakes
#          ↓
#     Final prediction
# so here learning_rate = 0.1 , a tree only has smaller contribution in correcting the errors, 1.0 - larger contribution
# max_depth = depth of individual regressor tress, controls the complexity of the tress, both 3 and 4 are tested.
```

```python
random_search = RandomizedSearchCV(GradientBoostingRegressor(random_state = 42), param_distributions = param_grid, n_iter = 20, cv =5, scoring='r2', n_jobs = -1, verbose = 2)
# cv = 5 dividing the data into 5 folds, n_jobs = -1, take all the available fn from cpu, verbose = 2 - no need all steps
```

```python
random_search.fit(x_train, y_train)
```

```python
print('Best Parametrs:', random_search.best_params_)
```

```python
print('Best Score', random_search.best_score_)
```

* Best Parameters: Using 200 trees (n_estimators), a maximum depth of 4 (max_depth), and a learning rate of 0.1 resulted in the best performance.

* Best Score: This optimal combination achieved an R² Score of 0.8216 during the 5-fold cross-validation, confirming a significant improvement over the initial baseline score of 0.666

```python
#training optimized model
gb_best = random_search.best_estimator_
gb_best.fit(x_train, y_train)
y_pred_best = gb_best.predict(x_test)
```

```python
print('\n--Optimized model performace--')
print('mean squared error', mean_squared_error(y_test, y_pred_best))
print('mean absolute eeror', mean_absolute_error(y_test, y_pred_best))
print('r2 score', r2_score(y_test, y_pred_best))
```

* R² Score increased from the baseline 0.776 to 0.819 (an improvement of ≈4.3%).

* Mean Absolute Error (MAE) decreased from 0.372 to 0.327.

* Mean Squared Error (MSE) decreased from 0.294 to 0.238.
