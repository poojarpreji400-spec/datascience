```python
data['belongs_to_collection'] = data['belongs_to_collection'].fillna('no data')#, inplace = True)
data['homepage'] = data['homepage'].fillna('no website')#,inplace = True) , no need of inplace because we are assigning values like data['homepage'] =...     
data['production_companies'] =  data['production_companies'].fillna('Notavailable')
data['production_countries'] = data['production_countries'].fillna('no data')
data['runtime'] = data['runtime'].fillna(data['runtime'].median())
data['spoken_languages'] = data['spoken_languages'].fillna('not available')    
data['tagline'] =  data['tagline'].fillna('no tagline')      
data['Keywords'] = data['Keywords'].fillna('no keywords')                                
data['crew'] = data['crew'].fillna('no data')       
data['cast'] = data['cast'].fillna('no information')
data['overview'] = data['overview'].fillna('no overview')
data['genres'] = data['genres'].fillna('no genres')
data['poster_path'] = data['poster_path'].fillna('no poster_path')
```

```python
#12.Extract the genres (note: some movies may have multiple genres, split them).
#13.Count the number of movies in each genres.
from collections import Counter # counter is used to count how many times an intem occur eg counter({drama,comedy,drama}) - drama = 2, comedy = 1
import ast #ast.literal_eval converts "[{'id': 18, 'name': 'Drama'}, {'id': 35, 'name': 'Comedy'}]" type of data to lists
def extract_genre_names(x):
    try:
        genres = ast.literal_eval(x) # first converting into list Eg [ {'id': 18, 'name': 'Drama'}, {'id': 35, 'name': 'Comedy'}]
        return [i['name'] for i in genres] # then only returning the genre name in the entire list of genres Eg ['drama', 'comedy']etc
    except (SyntaxError, ValueError):
        return ['Unknown']   #If a row contains bad/malformed data, instead of stopping the entire program, the function returns:
data['genres_list'] = data['genres'].apply(extract_genre_names)
all_genres = [genre for sublist in data['genres_list'] for genre in sublist] #how this works is given below, it converts the list to flat by putting all together
#['Drama', 'Comedy'] ─┐
#['Action']            ├──> ['Drama', 'Comedy', 'Action', 'Drama', 'Romance']
#['Drama', 'Romance'] ─┘
genre_counts = Counter(all_genres) # count all genres 
genres_counts_df = pd.DataFrame(genre_counts.items(), columns = ['Genre', 'count'])
print('genre_counts:')
for genre, count in genre_counts.items():
    print(f'{genre} :{count}')
```

```python
#14.Plot a bar chart of the movie counts per genre.
plt.figure(figsize = (12,6))
plt.bar(genres_counts_df['Genre'], genres_counts_df['count']) # instead of writing matplotlib.pyplot.bar() we give plt.bar
plt.xticks(rotation = 45)
plt.title('Movie Counts per Genre')
plt.xlabel('Genre')
plt.ylabel('Number of movies')
plt.show()
```

```python
#15.Sort movies by popularity (or rating/votes depending on dataset).
#16.List the Top 10 Most Popular Films with their popularity values.
top_10_popular_films = data.sort_values(by = 'popularity', ascending = False).head(10)
top_10_popular_films[['title','popularity']].head(10)
```

```python
#17.Plot a line chart showing these 10 films by popularity.
plt.figure(figsize = (10,6))
plt.plot(top_10_popular_films['title'],top_10_popular_films['popularity'],marker = 'o', linestyle = '--', color = 'green', linewidth = 2)
plt.xlabel('movie_title')
plt.ylabel('popularity')
plt.title('Top 10 popular movies')
plt.xticks(rotation = 45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
#18.Find the Top 7 highest budgeted films.
#19.List their Titles and Budgets.
top_7_revenue = data.sort_values(by = 'revenue', ascending = False)
top_7_revenue[['title', 'revenue']].head(7)
```

```python
#20.Plot a bar chart or histogram for these 7 films
#histogram
top_7_revenue = data.sort_values(by = 'revenue', ascending= False).head(7)
plt.figure(figsize = (11,5))
plt.hist(top_7_revenue['revenue'], bins = 7, color = 'blue', edgecolor = 'black')
plt.title('Top 7 budgeted movies')
plt.ylabel('Revenue')
plt.xlabel('Films', fontsize = 20)
plt.tight_layout()
plt.xticks(top_7_revenue['revenue'], top_7_revenue['title'],rotation = 45)
plt.show()
```
