```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
```

The price range form 0 - low cost, 1 - Medium cose, 2 - High Cost, 3 - Very High Cost

```python
x = data.drop('price_range',axis = 1)
y = data['price_range']
```

```python
x_train, x_test,y_train, y_test = train_test_split(x,y, test_size = 0.2, random_state = 42, stratify = y) # stratify reduces class imbalance
```

For example, if your original data is:

25% class 0, 25% class 1, 25% class 2, 25% class 3

then after splitting:

Train(80%) → approximately 25%, 25%, 25%, 25%

Test(20%) → approximately 25%, 25%, 25%, 25%

```python
scaler = MinMaxScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.fit_transform(x_test)
```

```python
model = RandomForestClassifier(n_estimators = 100, random_state = 42) # n_estimators = 100 - Create 100 decision trees and combine their predictions to make the final prediction.
model.fit(x_train_scaled, y_train)
```

Each tree has their own predictions

tree 1 - class 2
tree 2 - class 1
tree 3 - class 2
...
tree 100 - class 2

final votes -
class 1 - 10
class 2 - 60
class 3 - 15

Random Forest chooses majority voting one ie, class 2

```python
y_pred = model.predict(x_test_scaled)
```

```python
print('accuracy', accuracy_score(y_test, y_pred))
print('\nclassification report', classification_report(y_test, y_pred))
print('\nconfusion matrix\n', confusion_matrix(y_test, y_pred))
```

Class 0  - 96 is classified correctly as class 0 and 4 is incorrectly classifid as class 1
