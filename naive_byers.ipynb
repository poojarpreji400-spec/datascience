```python
import numpy as np 
import pandas as pd 
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_breast_cancer
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from sklearn.preprocessing import StandardScaler
```

```python
df = load_breast_cancer()
```

```python
data = pd.DataFrame(df.data, columns = df.feature_names) # converting data to data frame
data['target'] = df.target
```

```python
print('First 5 row od data/n', data.head())
```

```python
print('Dataset info\n', data.info())
```

```python
print('Missing values per column\n', data.isnull().sum())
```

```python
print('Target Value Counts:', data['target'].value_counts())
```

```python
x = data.drop('target', axis = 1)
y = data['target']
```

```python
scaler = StandardScaler()
```

```python
x_scaled = scaler.fit_transform(x)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x_scaled, y, test_size = 0.2, random_state = 42)
```

```python
print('Data preprocessing completed')
print('Training set shape', x_train.shape)
print('Testing set shape', x_test.shape)
```

```python
model = GaussianNB()
```

```python
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
print('Evation Model Results\n Accuracy Score:', round(accuracy_score(y_test, y_pred)*100, 2),'%')
# converts decimal into percentage using *100 and then rounding of to 2, and adding % symbol
print('confussion matrix:',confusion_matrix(y_test, y_pred))
print('classification matrix:', classification_report(y_test, y_pred))#, target_names = df.target_names))
```

```python
print('Evation Model Results\n Accuracy Score:', round(accuracy_score(y_test, y_pred)*100, 2),'%')
# converts decimal into percentage using *100 and then rounding of to 2, and adding % symbol
print('confussion matrix:',confusion_matrix(y_test, y_pred))
print('classification matrix:', classification_report(y_test, y_pred, target_names = df.target_names))# target names koduthilengil 0, 1 maatharme varu
```
