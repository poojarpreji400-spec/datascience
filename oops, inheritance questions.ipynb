```python
#✅ Question 4: Create three classes: Grandparent, Parent, and Child. Each class has one method: Grandparent: show_grandparent() Parent: show_parent()
#Child: show_child() The Parent inherits from Grandparent, and Child inherits from Parent. Create an object of Child and call all three methods.
class Grandparent():
    def show_grandparent(self):
        print('hi')
class Parent(Grandparent):
    def show_parent(self):
        print('hello')
class Child(Parent):
    def show_child(self):
        print('good')
c= Child()
c.show_grandparent()
c.show_parent()
c.show_child()
```

```python
#✅ Question 5: Create three classes: Animal, Mammal, and Dog. Each should have a method: Animal: eat() Mammal: walk() Dog: bark() 
#Create a Dog object and call all the methods.
class Animal():
    def eat(self):
        print('eats')
class Mammal(Animal):
    def walk(self):
        print('walks')
class Dog(Mammal):
    def bark(self):
        print('barks')
d= Dog()
d.eat()
d.walk()
d.bark()
```

```python
#✅ Question 6: Create three classes: Device with method device_info() → prints "General device"
#Laptop inherits from Device and has method laptop_info() → prints "This is a laptop" 
#GamingLaptop inherits from Laptop and has method gaming_info() → prints "High performance for gaming" Create an object of GamingLaptop and call all methods.
class Device():
    def device_info(self):
        print('General device')
class Laptop(Device):
    def laptop_info(self):
        print('This is a laptop')
class GamingLaptop(Laptop):
    def gaming_info(self):
        print('High performance for gaming')
g= GamingLaptop()
g.device_info()
g.laptop_info()
g.gaming_info()
```

```python
#✅ Question 1:Animal Sounds Q: Create a class Animal with a method speak(). Create two subclasses: Cow and Sheep. Override speak() in each to print: "Cow says Moo"
#"Sheep says Baa"
class Animal():
    def speak(self):
        print('speaks')
class Cow(Animal):
    def speak(self):
        print('Cow says Moo')
class Sheep(Animal):
    def speak(self):
        print('Sheep says Baa')
c= Cow()
s= Sheep()
c.speak()
s.speak()
```

```python
#Create a class Shape with a method perimeter(). Create subclasses Square and Triangle and override perimeter(): Square: 4 * side Triangle: side1 + side2 + side3
#Use sample values: Square side = 5 Triangle sides = 3, 4, 5
class Shape():
    def perimeter(self):
        print('perimeters')
class Square(Shape):
    def perimeter(self):
        return 4 * side
class Triangle(Shape):
    def perimeter(self):
        return side1 + side2 + side3
side = 5
side1 = 3
side2 = 4
side3 = 5
s = Square()
t = Triangle()
print(s.perimeter())
print(t.perimeter())
```
