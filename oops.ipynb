```python
from abc import ABC, abstractmethod
class vehicle(ABC):
    @abstractmethod
    def start(self):
        pass
    def stop(self):
        pass
    def breaks(self):
        pass
class car(vehicle):
    def start(self):
        print('car is started')
    def stop(self):
        print('car is stopped')
    def breaks(self):
        print('car is breaked')
class bike(vehicle):
    def start(self):
        print('bike is started')
    def stop(self):
        print('bike is topped')
C = car()
C.stop()
C.start()
C.breaks()
B= bike()
B.stop()
B.start()
```

Single level inheritance

```python
class Animal():
    def speak(self):
        print('animal speak')
class dog(Animal):
    def talk(self):
        print('dog talk')
c= Animal()
c.speak()   
d = dog()
d.speak()
d.talk()
```

multilevel inheritance

```python
class Animal():
    def speak(self):
        print('animal speak')
class dog(Animal):
    def talk(self):
        print('dog talk')
class cat(dog):
    def meows(self):
        print('cat meows')
c= Animal()
c.speak()   
d = dog()
d.speak()
d.talk()
c = cat()
c.meows()
c.speak()
c.talk()
```

multiple level inheritance

```python
class father:
    def talk(self):
        print('father talks')
class mother:
    def speaks(self):
        print('mother speaks')
class child(father, mother):
    def smiles(self):
        print('child smiles')
f = father()
f.talk()
m = mother()
m.speaks()
c= child()
c.talk()
c.speaks()
c.smiles()
```

Heirarchichal inheritance

```python
class Animal():
    def speak(self):
        print('animal speak')
class dog(Animal):
    def bark(self):
        print('dog speak')
class cat(Animal):
    def meows(self):
        print('cat meows')
a = Animal()
a.speak()
d= dog()
d.bark()
d.speak()
c= cat()
c.meows()
c.speak()
```

Hybrid Inheritance

```python
class animal():
    def speak(self):
        print('animal speak')
class bird:
    def chirp(self):
        print('bird chirp')
class dog(animal):
    def bark(self):
        print('dog bark')
class cat(dog):
    def meow(self):
        print('cat meows')
class ecosystem(animal,bird):
    def green(self):
        print('ecosytem consits bird and animal')
class jiraffe(animal):
    def tall(self):
        print('jiraffe tall')
a = animal()
a.speak()
d= dog()
d.speak()
d.bark()
c = cat()
c.meow()
c.speak()
b = bird()
b.chirp()
e= ecosystem()
e.speak()
e.chirp()
e.green()
j = jiraffe()
j.speak()
j.tall()
```
