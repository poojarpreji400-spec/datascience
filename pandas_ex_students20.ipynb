```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('students_20.csv')
data
```

```python
#Q1. Show the first 5 rows of the DataFrame.
print(data.head())
```

```python
#Q2. Show only the "Student Name" and "Math" columns.
a = print(data[['Student Name', 'Math']])
```

```python
#Q3. Find the average marks in each subject.
b = data[[  'Math','Science','English']].mean() # use double brackets for more than 1 column
print(b)
```

```python
plt.figsize = (4,10)
plt.plot(b.index, b.values, marker = 'o', linestyle = '--')
plt.xlabel('subjects')
plt.ylabel('avg')
plt.title('average marks in each subject')
plt.grid(axis = 'y')
plt.ylim(75,87)
plt.show()
```

```python
#Q4. Which student scored the highest in Science?
c = data['Science'].max()
print(c) 
```

```python
c1 = data.loc[data['Science'].idxmax()]
print(c1)   #returns the index (row label) of the first occurrence of the maximum value in a Series or DataFrame
```

```python
top_science = data[data['Science']== c]
print(top_science)   #df["Science"].max() → gives the highest Science mark.
# df[df["Science"] == max_science] → filters the row(s) where Science = max.
```

```python
c2 = data.loc[data['Science'] == c]
print(c2)
```

```python
c3 = data.loc[data['Science']. idxmax() , 'Student Name']
print(c3)
```

```python
#Q15. Create a new column "Grade" based on Total Marks. If Total ≥ 260 → "A"
#If 240 ≤ Total < 260 → "B" , Else → "C"
data['grade'] = pd.cut(data['total'], bins=[0,239,259,300],labels =['C','B','A'])#pd.cut() divides numerical values into groups (bins).0 to 239 → Grade C
#240 to 259 → Grade B
#260 to 300 → Grade A
print(data[['Student Name','grade','total']])
```

```python
#16. Show the average marks per subject for each Grade.
k = data.groupby('grade').mean(numeric_only = True)  #Groups rows with the same value
print(k)
```

```python
k2 = data.groupby('grade')[['English','Math','Science']].mean()
k2
```

```python
#Q17. Show students who have the same marks in any two subjects.
print(data[(data['Math'] == data['Science']) | (data['Math'] == data['English']) | (data['English'] == data['Science'])])
```
