```python
import pandas as pd
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC # svm - support vector machine , svc - support vector classification
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
iris = datasets.load_iris()
```

```python
df = pd.DataFrame(data = iris.data, columns = iris.feature_names)
```

```python
df
```

```python
df['target'] = iris.target # add a column named target to the dataset
df['target_name'] = iris.target_names[iris.target] # add a column name called target_name to the dataset that contains the actual name corresponding to those numbers.
df # general syntax - df['column_name'] = values
```

```python
x= iris.data
y = iris.target
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

```python
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.fit_transform(x_test)
```

```python
svm_model = SVC(kernel ='rbf' , C = 1.0, gamma = 'scale') #kernal = 'rbf - a data cannot be always seperated by straight line because the data may be mixed together so in that caxes we uses rbf to create a non-linear decision boundary.
# c - determines how strightly the model tries to classify the training data points correctly. small c - little mistakes is fine, larger(simple boundary) c - tries harder to avoid mistakes(comples boundary), c = 1.0 is the default
# gamma - controls how much influence the datapoints have(over prediction). higher gamma - each datapoint have local influence(complex boundary), lowe gamma - each datapoint have wider influence(smoother boundary)
svm_model.fit(x_train_scaled, y_train)
```

Setosa is linearly separable.

Versicolor and Virginica overlap slightly — not fully separable by a straight line.
Hence, the RBF kernel allows the SVM to form a curved decision boundary that correctly separates these two species.

```python
y_pred = svm_model.predict(x_test_scaled)
```

```python
print('Training accuracy:', svm_model.score(x_train_scaled, y_train))
print('Test Accuracy:', svm_model.score(x_test_scaled, y_test))
```

```python
print('accuracy score', accuracy_score(y_test, y_pred))
print('confusion matrix\n', confusion_matrix(y_test, y_pred))
print('classification report\n', classification_report(y_test, y_pred))
```

```python
test_results = pd.DataFrame(scaler.inverse_transform(x_test_scaled),columns = iris.feature_names)
test_results['actual flower'] = iris.target_names[y_test] # here y_test contains only target name(because traget column), but we need target name so we used this code.
#this add a new column names actual flowet
test_results['predicted_flower'] = iris.target_names[y_pred]
```

```python
print(test_results) #To see each test(only contains test data) flower's original measurements, its actual species, and the species predicted by your SVM model.
```

```python
import joblib # joblib - python library to save and load machine- learning models.
joblib.dump(svm_model, 'sm_model.pkl') #svm_model - my trained model , svm_model.pkl - model is saved as pkl(python pickle file)
joblib.dump(scaler, 'scaler,pkl') # scaler - the scaler i used above to scale the x_train, test, scaler.pkl - file is saved there
```
